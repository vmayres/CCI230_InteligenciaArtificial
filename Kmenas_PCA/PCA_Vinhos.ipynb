```python
from sklearn.datasets import load_wine
import pandas as pd

vinhos = load_wine()
# criando DataFrame (Pandas):
df = pd.DataFrame(vinhos.data, columns=vinhos.feature_names)
df['target'] = vinhos.target
df
```

```python
import seaborn as sns
import pandas as pd

sns.pairplot(df, hue='target', palette="deep")
```

```python
sns.set(rc={'figure.figsize':(12,8)})
sns.heatmap(df.iloc[:,:-1].corr(), annot=True)
```

```python
vinhos, regiao = load_wine(return_X_y=True)

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
padronizado = scaler.fit_transform(vinhos) # 13 entradas, corrigir escala

pca = PCA(n_components=2)
dados = pca.fit_transform(padronizado)
print(len(dados))

sns.scatterplot(x=dados[:,0], y =dados[:,1], hue=regiao, palette="deep")
```

```python
pca = PCA(n_components=3)
dados = pca.fit_transform(padronizado)

import plotly.express as px

px.scatter_3d(x=dados[:,0], y=dados[:,1], z=dados[:,2], color=regiao)
```
