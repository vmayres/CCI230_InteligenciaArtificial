# PCA Penguins Dataset

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.preprocessing import LabelEncoder

import seaborn as sns

# Carrengo o dataset
penguin = sns.load_dataset("penguins")
penguin.info()

#criando dataframe
df = penguin.dropna()

# separa entrada e saída
entrada = df[['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']]
saida = df['species']
```

#### Criando matrxi de graficos corelaciando as atributos dos pinguins.



```python
sns.pairplot(df, hue='species', palette="deep")
```

#### Matrix de confusao ou Heatmap

```python
# Encode the 'species' column
label_encoder = LabelEncoder()
df.loc[:, 'species_encoded'] = label_encoder.fit_transform(df['species'])

# Compute the correlation matrix without the original 'species' column and other non-numeric columns
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True)
```

#### Aplicao do modelp PCA 

utilizando 

```python
pca_model_2dim = PCA(n_components=2)
duas_dimensoes = pca_model_2dim.fit_transform(entrada)

sns.scatterplot(x=duas_dimensoes[:,0], y =duas_dimensoes[:,1], hue=saida, palette="deep")
```

```python
pca_model_3dim = PCA(n_components=3)
tres_dimensoes = pca_model_3dim.fit_transform(entrada)

px.scatter_3d(x=tres_dimensoes[:,0], y=tres_dimensoes[:,1], z=tres_dimensoes[:,2], color=saida)
```

```python
entrada_padrao = StandardScaler().fit_transform(entrada)

pca_model_2dim = PCA(n_components=2)
duas_dimensoes = pca_model_2dim.fit_transform(entrada_padrao)

sns.scatterplot(x=duas_dimensoes[:,0], y =duas_dimensoes[:,1], hue=saida, palette="deep")
```

```python
entrada_padrao = StandardScaler().fit_transform(entrada)

pca_model_3dim = PCA(n_components=3)
tres_dimensoes = pca_model_3dim.fit_transform(entrada)

px.scatter_3d(x=tres_dimensoes[:,0], y=tres_dimensoes[:,1], z=tres_dimensoes[:,2], color=saida)
```
